```python
import re
def get_groups_from_filenames(fpath):
    last_dir = os.path.basename(fpath)
    groups = set()
    for fname in os.listdir(fpath):
        if fname.endswith(".tif"):
            match = re.search(rf"{re.escape(last_dir)}(\d+)", fname)
            if match:
                no = match.group(1)
                groups.add(f"{last_dir[-1]}{no}")
    return sorted(list(groups))
```
