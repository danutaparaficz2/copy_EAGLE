```python
import os
import pickle
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
from sklearn.metrics import confusion_matrix
```

```python
results_dir = 'results'  
pickle_files = [f for f in os.listdir(results_dir) if f.endswith('.pkl') or f.endswith('.pickle')]

data = {}
for fname in pickle_files:
    name = fname.split('_')[0]  # Remove file extension for key
    file_path = os.path.join(results_dir, fname)
    with open(file_path, 'rb') as f:
        data[name] = data.get(name, [])
        data[name].append(pickle.load(f))
```

```python
all_preds = {}
all_true = {}

for name, items in data.items():
    preds_list, true_list = [], []
    #print(name, len(items))
    for item in items:
        #print(item['pred_labels'].shape, item['true_labels'].shape)
        preds_list.append(item['pred_labels'])
        true_list.append(item['true_labels'])

    all_preds[name] = np.concatenate(preds_list)
    all_true[name] = np.concatenate(true_list)
    print(name, all_preds[name].shape, all_true[name].shape)
```

```python
name = 'tiso'
preds = all_preds[name]
trues = all_true[name]
print(f"Results for {name}:")
n_cols = preds.shape[1] if preds.ndim > 1 else 1
for col in range(n_cols):
    col_preds = preds[:, col] if n_cols > 1 else preds
    col_trues = trues[:, col] if n_cols > 1 else trues
    print(f"  Column {col}:")
    print("    Accuracy:", accuracy_score(col_trues, col_preds))
    print("    Precision:", precision_score(col_trues, col_preds, average='weighted', zero_division=0))
    print("    Recall:", recall_score(col_trues, col_preds, average='weighted', zero_division=0))
    print("    F1 Score:", f1_score(col_trues, col_preds, average='weighted', zero_division=0))
    print(classification_report(col_trues, col_preds, zero_division=0))
    print("    Confusion Matrix:")
    print(confusion_matrix(col_trues, col_preds))
```
